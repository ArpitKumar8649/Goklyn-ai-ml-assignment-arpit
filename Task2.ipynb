```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. LLM: Gemini model via LangChain
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",   # fast & free tier friendly
    temperature=0.3
)

# 2. System + user prompt template
system_prompt = """
You are a helpful AI assistant for a Security Operations Center (SOC).
Your tasks:
- Analyze the user's question,
- Explain security / SIEM / SOC concepts clearly,
- Give step-by-step, beginner-friendly answers,
- When relevant, connect ideas to SIEM rules, alerts, and incident response.

If the question is not about security, still answer politely and helpfully.
Keep answers concise but clear.
"""

human_prompt = "{user_input}"

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", system_prompt),
        ("human", human_prompt),
    ]
)

# 3. Chain = Prompt -> Gemini -> Plain text
chain = prompt | llm | StrOutputParser()
```

```python
def run_agent():
    print("Security Assistant Agent (LangChain + Gemini)")
    print("Type 'exit' or 'quit' to stop.\n")

    while True:
        user_input = input("You: ")
        if user_input.lower() in ["exit", "quit", "q"]:
            print("Agent: Goodbye! 👋")
            break

        # Invoke the LangChain chain
        response = chain.invoke({"user_input": user_input})
        print("\nAgent:", response, "\n")

# Run the agent
run_agent()
```
